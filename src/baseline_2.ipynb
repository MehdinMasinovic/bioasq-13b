### Step 2: Consume E-utilities API to get relevant documents

```python
import requests

def get_most_relevant_documents(query, max_results=10, api_key="44c6e8c525237134d87515d972ff3ad85608"):
    base_url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esearch.fcgi"
    params = {
        "db": "pubmed",
        "term": query,
        "retmode": "json",
        "retmax": max_results,
        "api_key": api_key
    }
    response = requests.get(base_url, params=params)
    response.raise_for_status()
    data = response.json()
    return data["esearchresult"]["idlist"]
```

## Fetch Titles and Abstracts for PubMed IDs

```python
import requests
import xml.etree.ElementTree as ET

def fetch_pubmed_details(id_list):
    base_url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi"
    ids = ",".join(id_list)
    params = {
        "db": "pubmed",
        "id": ids,
        "retmode": "xml"
    }
    response = requests.get(base_url, params=params)
    response.raise_for_status()
    articles = []
    root = ET.fromstring(response.content)
    for article in root.findall(".//PubmedArticle"):
        pmid = article.findtext(".//PMID")
        title = article.findtext(".//ArticleTitle") or ""
        abstract = " ".join([abst.text or "" for abst in article.findall(".//AbstractText")])
        articles.append({
            "pmid": pmid,
            "title": title,
            "documentAbstract": abstract
        })
    return articles
  
   
```

```python
from rank_bm25 import BM25Okapi
from nltk.tokenize import word_tokenize

for question in bioasq_13b_questions:
    doc_ids = question['documents_api']
    if not doc_ids:
        print(f"Skipping question `{question['id']}` due to no documents found.\n")
        continue

    # Fetch document details
    docs = fetch_pubmed_details(doc_ids)
    if not docs:
        print(f"No details fetched for question `{question['id']}`.\n")
        continue

    # Prepare documents for BM25
    full_doc = [doc["title"] + " " + doc["documentAbstract"] for doc in docs]
    tokenized_docs = [word_tokenize(doc.lower()) for doc in full_doc]
    tokenized_question = word_tokenize(question['body'].lower())

    bm25 = BM25Okapi(tokenized_docs)
    scores = bm25.get_scores(tokenized_question)
    ranked_docs = sorted(zip(docs, scores), key=lambda x: x[1], reverse=True)
    top_docs = ranked_docs[:10]

    print(f"Top articles for question `{question['id']}`:")
    for doc, score in top_docs:
        print(f"\nDocument ID: {doc['pmid']}")
        print(f"Title: {doc['title']}")
        print(f"Score: {score}")
        print(f"Abstract: {doc['documentAbstract']}\n")
    print("="*60)
```

### Step 3: Rank documents with "Traditional IR" model

```python
from rank_bm25 import BM25Okapi
from nltk.tokenize import word_tokenize

for question in bioasq_13b_questions:
    doc_ids = question['documents_api']
    if not doc_ids:
        print(f"Skipping question `{question['id']}` due to no documents found.\n")
        continue

    # Fetch document details
    docs = fetch_pubmed_details(doc_ids)
    if not docs:
        print(f"No details fetched for question `{question['id']}`.\n")
        continue

    # Prepare documents for BM25
    full_doc = [doc["title"] + " " + doc["documentAbstract"] for doc in docs]
    tokenized_docs = [word_tokenize(doc.lower()) for doc in full_doc]
    tokenized_question = word_tokenize(question['body'].lower())

    bm25 = BM25Okapi(tokenized_docs)
    scores = bm25.get_scores(tokenized_question)
    ranked_docs = sorted(zip(docs, scores), key=lambda x: x[1], reverse=True)
    top_docs = [doc['pmid'] for doc, score in ranked_docs[:10]]

    print(f"Ranked documents for question `{question['id']}`:")
    for doc, score in ranked_docs[:10]:
        print(f"Document ID: {doc['pmid']}, Score: {score}")
    print("\n")
```
